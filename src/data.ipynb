```python
#combine datasets
raw_path = "../data/raw"
files = glob.glob(raw_path + "/*.csv")
total_raw_data = []
for file in files:
    df = pd.read_csv(file)
    total_raw_data.append(df)

data = pd.concat(total_raw_data, ignore_index=True)
pd.set_option('display.max_columns', None)
```
