```python
#load clean and process data
# 
#  
# DATA FROM https://github.com/JeffSackmann/tennis_atp 

# surface
# tourney_level
# best_of
# rank_diff
# rank_points_diff
# age_diff
# height_diff
# seed_diff
# entry_diff


import numpy as np
import pandas as pd 
import glob
```

```python

raw_path = "../data/raw"
files = glob.glob(raw_path + "/*.csv")
total_raw_data = []
for file in files:
    df = pd.read_csv(file)
    total_raw_data.append(df)
```

```python
dfs = pd.concat(total_raw_data, ignore_index=True)
pd.set_option('display.max_columns', None)
```

```python
dfs = dfs.dropna(subset=[
    'winner_rank',
    'loser_rank',
    'winner_rank_points',
    'loser_rank_points'
])

numeric_cols = [
    'winner_age', 'loser_age',
    'winner_ht', 'loser_ht',
    'winner_seed', 'loser_seed'
]

for col in numeric_cols:
    dfs[col] = dfs[col].fillna(dfs[col].median())

dfs['winner_entry'] = dfs['winner_entry'].fillna('UNK')
dfs['loser_entry'] = dfs['loser_entry'].fillna('UNK')

dfs = dfs.dropna(subset=['surface', 'tourney_level'])
```

```python
dfs['rank_diff'] = dfs['winner_rank'] - dfs['loser_rank']


dfs['rank_points_diff'] = dfs['winner_rank_points'] - dfs['loser_rank_points']
dfs['age_diff'] = dfs['winner_age'] - dfs['loser_age']
dfs['height_diff'] = dfs['winner_ht'] - dfs['loser_ht']
dfs['seed_diff'] = dfs['winner_seed'] - dfs['loser_seed']

surfaces = {
    'Hard': 0,
    'Clay': 1,
    'Grass': 2
}

dfs['surface'] = dfs['surface'].map(surfaces)

dfs['Player1_id'] = dfs['winner_id']
dfs['Player2_id'] = dfs['loser_id']
    
ys= dfs[['winner_id']]

xs = dfs[[
    'Player1_id',
    'Player2_id',
    'surface',
    'best_of',
    'rank_diff',
    'rank_points_diff',
    'age_diff',
    'height_diff',
    'seed_diff',
]]

xs
```

```python

xs.to_csv('../data/processed/tennis_features.csv', index=False)
ys.to_csv('../data/processed/tennis_labels.csv', index=False)
```
