```python
# DATA FROM https://github.com/JeffSackmann/tennis_atp 



import numpy as np
import pandas as pd 
import glob
```

```python

raw_path = "../data/raw"
files = glob.glob(raw_path + "/*.csv")
total_raw_data = []
for file in files:
    df = pd.read_csv(file)
    total_raw_data.append(df)
```

```python
dfs = pd.concat(total_raw_data, ignore_index=True)
pd.set_option('display.max_columns', None)

dfs
```

```python

base_cols = [
    # IDs & ranks
    "winner_id", "loser_id", "winner_rank", "loser_rank", "winner_rank_points", "loser_rank_points",
    # Biographical
    "winner_age", "loser_age", "winner_ht", "loser_ht",
    # In-match stats
    "w_ace","w_df","w_svpt","w_1stIn","w_1stWon","w_2ndWon","w_SvGms","w_bpSaved","w_bpFaced",
    "l_ace","l_df","l_svpt","l_1stIn","l_1stWon","l_2ndWon","l_SvGms","l_bpSaved","l_bpFaced",
    # Optional context
    "surface","tourney_level","tourney_date","round","best_of","minutes"
]
available = [c for c in base_cols if c in df.columns]
d = df[available].copy()

# Map to Player1 / Player2 features (winner perspective)
p1_cols = {
    "Player1_id": "winner_id",
    "Player2_id": "loser_id",
    "P1_rank": "winner_rank", "P2_rank": "loser_rank",
    "P1_rank_points": "winner_rank_points", "P2_rank_points": "loser_rank_points",
    "P1_age": "winner_age", "P2_age": "loser_age",
    "P1_ht": "winner_ht", "P2_ht": "loser_ht",
    "P1_ace":"w_ace","P1_df":"w_df","P1_svpt":"w_svpt","P1_1stIn":"w_1stIn","P1_1stWon":"w_1stWon",
    "P1_2ndWon":"w_2ndWon","P1_SvGms":"w_SvGms","P1_bpSaved":"w_bpSaved","P1_bpFaced":"w_bpFaced",
    "P2_ace":"l_ace","P2_df":"l_df","P2_svpt":"l_svpt","P2_1stIn":"l_1stIn","P2_1stWon":"l_1stWon",
    "P2_2ndWon":"l_2ndWon","P2_SvGms":"l_SvGms","P2_bpSaved":"l_bpSaved","P2_bpFaced":"l_bpFaced",
}
context_cols = [c for c in ["surface","tourney_level","tourney_date","round","best_of","minutes"] if c in d.columns]
p1_map = {**p1_cols, **{c:c for c in context_cols}}

p1 = d.rename(columns=p1_map)
p1["label"] = 1

# Mirror (loser perspective)
p2_cols = {k.replace("1","X").replace("2","1").replace("X","2"):v for k,v in p1_cols.items()}  # swap P1<->P2
p2_map = {**p2_cols, **{c:c for c in context_cols}}
p2 = d.rename(columns=p2_map)
p2["label"] = 0

sym = pd.concat([p1, p2], ignore_index=True)
```

```python


dfs = dfs.copy()

num_cols = [
    "winner_rank", "loser_rank",
    "winner_rank_points", "loser_rank_points",
    "winner_age", "loser_age",
    "winner_ht", "loser_ht",
    "winner_seed", "loser_seed",
    "best_of"
]
for c in num_cols:
    if c in dfs.columns:
        dfs[c] = pd.to_numeric(dfs[c], errors="coerce")

if "winner_seed" in dfs.columns:
    dfs["winner_seed"] = dfs["winner_seed"].fillna(0)
if "loser_seed" in dfs.columns:
    dfs["loser_seed"] = dfs["loser_seed"].fillna(0)

surface_map = {"Hard": 0, "Clay": 1, "Grass": 2}
if "surface" in dfs.columns:
    dfs["surface_mapped"] = dfs["surface"].map(surface_map).fillna(-1).astype(int)
else:
    dfs["surface_mapped"] = -1

dfs["rank_diff"]        = dfs["winner_rank"]        - dfs["loser_rank"]
dfs["rank_points_diff"] = dfs["winner_rank_points"] - dfs["loser_rank_points"]
dfs["age_diff"]         = dfs["winner_age"]         - dfs["loser_age"]
dfs["height_diff"]      = dfs["winner_ht"]          - dfs["loser_ht"]
dfs["seed_diff"]        = dfs["winner_seed"]        - dfs["loser_seed"]

dfs["Player1_id"] = dfs["winner_id"]
dfs["Player2_id"] = dfs["loser_id"]
dfs["label"] = 1

feature_cols = [
    "Player1_id",
    "Player2_id",
    "surface_mapped",
    "best_of",
    "rank_diff",
    "rank_points_diff",
    "age_diff",
    "height_diff",
    "seed_diff",
    "label",
]

p1 = dfs[feature_cols].copy()  # P1 = winner rows

# 7) Mirror: swap IDs and flip signs of diffs, set label = 0 (P1 = loser)
p2 = p1.copy()
p2["Player1_id"], p2["Player2_id"] = p1["Player2_id"], p1["Player1_id"]
for c in ["rank_diff", "rank_points_diff", "age_diff", "height_diff", "seed_diff"]:
    p2[c] = -p1[c]
p2["label"] = 0

balanced = pd.concat([p1, p2], ignore_index=True)
balanced = balanced.sample(frac=1.0, random_state=42).reset_index(drop=True)

X = balanced.drop(columns=["label"])
Y = balanced["label"].astype(int)

#X_model = X.drop(columns=["Player1_id", "Player2_id"])
# Simple imputation for any remaining numeric NaNs
X = X.fillna(X.median(numeric_only=True))

X_model
```
